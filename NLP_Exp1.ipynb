## **Types of tokenization**
1.whitespace and word tokenization :
we are using split for whitespace.This tokenizes the string purely based on whitespace characters.Notice how punctuation marks are kept attached to the words.
In word_tokenization Notice how punctuation marks (like ,, !, ?, ‘, and ’) are separated into their own distinct tokens, which is much better for natural language processing tasks.

```python
import nltk
from nltk.tokenize import word_tokenize


nltk.download('punkt_tab')

text = """
        Hey everyone!!! 😍 Welcome to my travel blog, @TravelWithMe! ✈️🌍
        Follow me on Instagram (@travel_diary), X (@explore_world), and Facebook (facebook.com/travelwithme).
        Today, I'm exploring amazing destinations like Goa, Manali, Kashmir, and Bali!!! Have you ever visited these places?
        🤔 Share your experiences using #TravelGoals, #Wanderlust, and #IncredibleIndia.
        Visit my website at https://www.travelwithme.com or email me at hello@travelwithme.com.
        For bookings, contact +91-98765-43210!!! Don't forget to like, comment, share & subscribe...
        😊 Is this your dream vacation? Let's explore the world together!!! 🚀
        """

#simple tokenization
token = text.split()
print(token)
print("total words=",len(token))

#using word tokenization
word = word_tokenize(text)
print(word)
print("total words=",len(word))
```

## **2.sentence tokenozation**
Spacy and sent has different number of sentences as output.

```python
from nltk.tokenize import sent_tokenize
import spacy
nlp = spacy.load('en_core_web_sm')

sentence = sent_tokenize(text)
print(sentence)
print("total sentemces=",len(sentence))

doc = nlp(text)
sentences = [sent.text for sent in doc.sents]
print("         ")
print("----------USING SPACY-----------")
print(sentences)
print("total sentences=",len(sentences))
```

# **Rule based** **tokenization**
Regular expression tokenization

```python
import nltk
from nltk.tokenize import RegexpTokenizer
import re


tokenizer = RegexpTokenizer(r"[\w']+")
tokens = tokenizer.tokenize(text)
print(           )
print("-----Regular Expression-----")
print(tokens)
print("total tokens=",len(tokens))

# re.findall returns the list of tokens directly
tokens = re.findall(r"\w+(?:'\w+)*|[^\w\s]", text)
print(           )
print("-----Regular Expression including punctuation-----")
print(tokens)
print("total tokens=",len(tokens))
```

# **Social Media Tokenization**
splitting social media text into meaningful tokens while handling hashtags (#), mentions (@), emojis, URLs, and abbreviations.
Here:
- reduce_len=True reduces excessively repeated characters.
- strip_handles=False keeps user mentions.
- preserve_case=True retains uppercase and lowercase letters as written.

```python
from nltk.tokenize import TweetTokenizer


tokenizer = TweetTokenizer()
tokens = tokenizer.tokenize(text)
print(tokens)



tokenizer = TweetTokenizer(preserve_case=True,strip_handles=False,reduce_len=True)
tokens = tokenizer.tokenize(text)
print("     ")
print("---------Emoji-------")
print(tokens)
```

# **Domain Specific tokenization**
splitting text into meaningful tokens based on a particular field or domain, such as medical, legal, financial, or computer science.

```python
#Medical Domain Area

texts = "The patient with high blood pressure was advised to monitor their blood pressure regularly after recovering from COVID-19."

medical_terms = ["blood pressure","COVID-19","high blood pressure"]

medical_terms.sort(key=len, reverse=True)

domain_pattern = "|".join(re.escape(term) for term in medical_terms)

pattern = rf"{domain_pattern}|\b\w+\b|[^\w\s]"
tokens = re.findall(pattern, texts,flags=re.IGNORECASE)
print(tokens)
print("total tokens=",len(tokens))
```

# Stemming
1.   Porter Stemming
2.   Snowball Stemming
3.   Lancaster Stemming

```python
import nltk
from nltk.stem import PorterStemmer,SnowballStemmer,LancasterStemmer
nltk.download('punkt')

words = ["playing", "plays", "played", "studies", "caring","connection","happiness","relational"]

#1.Porter Stemmer
stemmer = PorterStemmer()

#2. Snowball Stemmer
ss = SnowballStemmer("english")

#3. Lancaster Stemmer
ls = LancasterStemmer()


print(f"{'Word':<10} -> {'Porter':<10} {'Snowball':<10} {'Lancaster':<10}")
print("-" * 50)
for word in words:
    print(f"{word:<10} -> {stemmer.stem(word):<10} {ss.stem(word):<10} {ls.stem(word):<10}")
```

# Lemmitization

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()

words = ["playing", "plays", "played", "studies", "caring","connection","happiness","relational"]

print(f"{'Word':<10} -> {'Noun Lem':<15} {'Verb Lem':<15}")
print("-" * 50)

for word in words:
  print(f"{word:<10} -> {lemmatizer.lemmatize(word):<15} {lemmatizer.lemmatize(word, pos='v'):<15}")

#Spacy
print("\nSpaCy Lemmatization")
print("-" * 30)
doc = nlp(" ".join(words))

for token in doc:
    print(token.text, "->", token.lemma_)
```

# POS Tagging

```python
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

text = "The children are playing happily in the beautiful gardens while their parents were watching them."

words = word_tokenize(text)
tags = nltk.pos_tag(words)
stemmer = PorterStemmer()
lemma = WordNetLemmatizer()


print(f"{'Word':<15}{'POS Tag':<15}{'Stem':<15} {'lemma':<15}")
print("-" * 60)

for word , tag in tags :
   print(f"{word:<15} {tag:<15} {stemmer.stem(word):<15}{lemma.lemmatize(word)}")
```
