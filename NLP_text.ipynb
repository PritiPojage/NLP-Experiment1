```python


import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk import pos_tag

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

nlp = spacy.load("en_core_web_sm")


text = """
The students are studying carefully.
They studied different subjects and enjoyed learning.
The teachers were teaching interesting lessons.
"""

# ------------------------------------------------------------
#  TOKENIZATION
# ------------------------------------------------------------

tokens = word_tokenize(text)

print("=" * 70)
print("1. TOKENIZATION")
print("=" * 70)

print(tokens)

# ------------------------------------------------------------
# 5. PORTER STEMMING
# ------------------------------------------------------------

porter = PorterStemmer()

porter_stems = []

for word in tokens:
    if word.isalpha():
        porter_stems.append(porter.stem(word))

print("\n" + "=" * 70)
print("2. PORTER STEMMER")
print("=" * 70)

print(porter_stems)

# ------------------------------------------------------------
# 6. SNOWBALL STEMMING
# ------------------------------------------------------------

snowball = SnowballStemmer("english")

snowball_stems = []

for word in tokens:
    if word.isalpha():
        snowball_stems.append(snowball.stem(word))

print("\n" + "=" * 70)
print("3. SNOWBALL STEMMER")
print("=" * 70)

print(snowball_stems)

# ------------------------------------------------------------
# 7. NLTK POS TAGGING
# ------------------------------------------------------------

nltk_pos = pos_tag(tokens)

print("\n" + "=" * 70)
print("4. NLTK POS TAGGING")
print("=" * 70)

for word, tag in nltk_pos:
    print(f"{word:15} -> {tag}")

# ------------------------------------------------------------
# 8. NLTK LEMMATIZATION WITH POS
# ------------------------------------------------------------

lemmatizer = WordNetLemmatizer()

nltk_lemmas = []

for word, tag in nltk_pos:

    if word.isalpha():

        # Convert NLTK POS to WordNet POS
        if tag.startswith('J'):
            pos = 'a'       # Adjective

        elif tag.startswith('V'):
            pos = 'v'       # Verb

        elif tag.startswith('N'):
            pos = 'n'       # Noun

        elif tag.startswith('R'):
            pos = 'r'       # Adverb

        else:
            pos = 'n'

        lemma = lemmatizer.lemmatize(word, pos=pos)

        nltk_lemmas.append(lemma)

print("\n" + "=" * 70)
print("5. NLTK LEMMATIZATION")
print("=" * 70)

print(nltk_lemmas)

# ------------------------------------------------------------
# 9. SPACY PROCESSING
# ------------------------------------------------------------

doc = nlp(text)

spacy_pos = []
spacy_lemmas = []

for token in doc:

    if token.is_alpha:
        spacy_pos.append((token.text, token.pos_))
        spacy_lemmas.append(token.lemma_)

# ------------------------------------------------------------
# 10. SPACY POS TAGGING
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("6. SPACY POS TAGGING")
print("=" * 70)

for word, tag in spacy_pos:
    print(f"{word:15} -> {tag}")

# ------------------------------------------------------------
# 11. SPACY LEMMATIZATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("7. SPACY LEMMATIZATION")
print("=" * 70)

print(spacy_lemmas)

# ------------------------------------------------------------
# 12. SIDE-BY-SIDE COMPARISON
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("8. COMPARISON OF RESULTS")
print("=" * 70)

print(f"{'Original':15} {'Porter':15} {'Snowball':15} {'NLTK Lemma':15} {'spaCy Lemma':15}")
print("-" * 75)

# Create clean alphabetic tokens
clean_tokens = [word for word in tokens if word.isalpha()]

for i, word in enumerate(clean_tokens):

    print(
        f"{word:15} "
        f"{porter_stems[i]:15} "
        f"{snowball_stems[i]:15} "
        f"{nltk_lemmas[i]:15} "
        f"{spacy_lemmas[i]:15}"
    )
```
