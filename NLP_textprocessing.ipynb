 **Import** **Files**

```python
import nltk
import spacy
import re
import pandas as pd

from nltk.tokenize import word_tokenize, sent_tokenize, WhitespaceTokenizer
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer ,WordNetLemmatizer
from nltk import pos_tag

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

nlp = spacy.load("en_core_web_sm")
```

```python
text='''
Believe in yourself—even when things seem impossible!
Every small step, every re-attempt, and every self-improvement brings
you closer to success; don’t underestimate your potential, never stop learning,
and keep moving forward... because unshakable confidence, unstoppable determination,
 and wholehearted effort can turn today’s challenges into tomorrow’s achievements!
 '''
print(text)
print("\nCharacter count:", len(text))
```

# 1.**Tokenization**

```python

print("\n1. Regular Expression Tokenization:")
tokens_regex = re.findall(r"\b\w+\b", text)
print(tokens_regex)
print("Total Token Count:", len(tokens_regex))

print("\n2. MANUAL TOKENIZATION")
manual_tokens = text.split()
print(manual_tokens)
print("Total Token Count:", len(manual_tokens))

print("\n3. Whitespace Tokenization:")
tokens_whitespace = text.split()
print("Tokens:", tokens_whitespace)
print("Total Token Count:", len(tokens_whitespace))

print("\n4. SENTENCE TOKENIZATION")
sentences = sent_tokenize(text)
for sentence in sentences:
    print(sentence)
    print("Total Token Count:", len(sentences))

print("\n5. Tokenization Using spaCy:")
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
tokens_spacy = [token.text for token in doc]
print("Tokens:", tokens_spacy)
print("Total Token Count:", len(tokens_spacy))

sentences_spacy = [sent.text for sent in doc.sents]
print("Sentences:", sentences_spacy)
print("Total Sentence Count:", len(sentences_spacy))
```

# 2. **Stemming**

```python
porter = PorterStemmer()
print("\n1. Porter Stemmer:")
porter_stems = [porter.stem(word) for word in tokens_whitespace]
print(porter_stems)

snowball = SnowballStemmer("english")
print("\n2. Snowball Stemmer:")
snowball_stems = [snowball.stem(word) for word in tokens_whitespace]
print(snowball_stems)

lancaster = LancasterStemmer()
print("\n3. Lancaster Stemmer:")
lancaster_stems = [lancaster.stem(word) for word in tokens_whitespace]
print(lancaster_stems)

print("\n")
print("\n*******Comparison of Stemming Algorithms***********")
print("-" * 70)
print(f"{'Original Word':<18}{'Porter':<18}{'Snowball':<18}{'Lancaster':<18}")
print("-" * 72)

for word in tokens_whitespace:
    print(
        f"{word:<18}"
        f"{porter.stem(word):<18}"
        f"{snowball.stem(word):<18}"
        f"{lancaster.stem(word):<18}"
    )

```

# 3.**Lemmitization**

```python
nlp = spacy.load("en_core_web_sm")
lemmatizer = WordNetLemmatizer()
print("\n1. NLTK Lemmatization:")
print("-" *72)
lemmas = [lemmatizer.lemmatize(word) for word in tokens_whitespace]
print(lemmas)

doc = nlp(text)
print("\n2.spaCy Lemmatization:")
print("-" *72)
for token in doc:
    if not token.is_space:
       print(f"{token.text:<18} -> {token.lemma_}")
lemmas = [token.lemma_ for token in doc]

print("\nLemmatized Tokens:")
print("-" *72)
print(lemmas)
print("Total Lemmatized Tokens:", len(lemmas))
```

## **4.POS,Stem and Lemma **

```python

print("\n6. POS TAGGING USING NLTK")
print("-" * 72)

tokens_pos = tokens_whitespace(text)
pos_tags = pos_tag(tokens_pos)

print(f"{'Word':<20}{'POS Tag':<20}{'Description':<30}")
print("-" * 72)

pos_descriptions = {
    "NN": "Noun (singular)",
    "NNS": "Noun (plural)",
    "NNP": "Proper noun (singular)",
    "NNPS": "Proper noun (plural)",
    "VB": "Verb (base form)",
    "VBD": "Verb (past tense)",
    "VBG": "Verb (gerund/present participle)",
    "VBN": "Verb (past participle)",
    "VBP": "Verb (present, non-3rd person)",
    "VBZ": "Verb (3rd person singular)",
    "JJ": "Adjective",
    "JJR": "Adjective (comparative)",
    "JJS": "Adjective (superlative)",
    "RB": "Adverb",
    "RBR": "Adverb (comparative)",
    "RBS": "Adverb (superlative)",
    "DT": "Determiner",
    "IN": "Preposition/subordinating conjunction",
    "PRP": "Personal pronoun",
    "PRP$": "Possessive pronoun",
    "CC": "Coordinating conjunction",
    "CD": "Cardinal number",
    "MD": "Modal verb",
    "RP": "Particle",
    "TO": "To",
    "UH": "Interjection"
}

for word, tag in pos_tags:
    description = pos_descriptions.get(tag, "Other grammatical category")
    print(f"{word:<20}{tag:<20}{description:<30}")


# ==============================
# 7. POS TAGGING USING spaCy
# ==============================

print("\n7. POS TAGGING USING spaCy")
print("-" * 72)

doc = nlp(text)

print(f"{'Word':<20}{'POS':<15}{'Detailed POS':<20}{'Lemma':<20}")
print("-" * 75)

for token in doc:
    if not token.is_space:
        print(
            f"{token.text:<20}"
            f"{token.pos_:<15}"
            f"{token.tag_:<20}"
            f"{token.lemma_:<20}"
        )


# ==============================
# 8. POS, STEMMING AND LEMMA COMPARISON
# ==============================

print("\n8. POS, STEMMING AND LEMMA COMPARISON")
print("-" * 100)

porter = PorterStemmer()

print(
    f"{'Word':<18}"
    f"{'POS':<12}"
    f"{'Porter Stem':<18}"
    f"{'spaCy Lemma':<18}"
)

print("-" * 66)

for token in doc:
    if token.is_alpha:
        print(
            f"{token.text:<18}"
            f"{token.pos_:<12}"
            f"{porter.stem(token.text):<18}"
            f"{token.lemma_:<18}"
        )
```
